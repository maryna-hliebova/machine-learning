### Домашні завдання на відпрацювання поняття функції та її похідної

1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

x = sp.Symbol("x")

f = 10 * x - 5
f_derivative = sp.diff(f, x)

print(f"f(x) = {f}")
print(f"f'(x) = {f_derivative}")

x_values = np.linspace(0, 10, 100)

y_values = 10 * x_values - 5
derivative_values = np.full_like(x_values, float(f_derivative))

plt.plot(x_values, y_values, label="f(x) = 10x - 5")
plt.plot(x_values, derivative_values, label="f'(x) = 10", linestyle="--")

plt.title("Функція та її похідна")
plt.xlabel("Кількість годин навчання")
plt.ylabel("Оцінка")
plt.grid(True)
plt.legend()
plt.show()

print("Висновок: кожна додаткова година навчання збільшує оцінку на 10 балів.")
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
sin_function = sp.sin(x)
ln_function = sp.log(x)

sin_derivative = sp.diff(sin_function, x)
ln_derivative = sp.diff(ln_function, x)

print(f"f(x) = {sin_function}")
print(f"f'(x) = {sin_derivative}")
print(f"g(x) = {ln_function}")
print(f"g'(x) = {ln_derivative}")

x_sin_values = np.linspace(-2 * np.pi, 2 * np.pi, 100)
y_sin_values = np.sin(x_sin_values)
y_sin_derivative_values = np.cos(x_sin_values)

x_ln_values = np.linspace(0.1, 10, 100)
y_ln_values = np.log(x_ln_values)
y_ln_derivative_values = 1 / x_ln_values

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(x_sin_values, y_sin_values, label="f(x) = sin(x)")
plt.plot(x_sin_values, y_sin_derivative_values, label="f'(x) = cos(x)", linestyle="--")
plt.title("Функція sin(x) та її похідна")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(x_ln_values, y_ln_values, label="g(x) = ln(x)")
plt.plot(x_ln_values, y_ln_derivative_values, label="g'(x) = 1/x", linestyle="--")
plt.title("Функція ln(x) та її похідна")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
f = x ** 2 - 5 * x + 1
f_derivative = sp.diff(f, x)

print(f"f(x) = {f}")
print(f"f'(x) = {f_derivative}")

x_values = np.linspace(0, 100, 200)

y_values = x_values ** 2 - 5 * x_values + 1
derivative_values = 2 * x_values - 5

plt.figure(figsize=(10, 6))

plt.plot(x_values, y_values, label="f(x) = x² - 5x + 1")
plt.plot(x_values, derivative_values, label="f'(x) = 2x - 5", linestyle="--")

plt.title("Функція витрат компанії та її похідна")
plt.xlabel("Кількість вироблених одиниць")
plt.ylabel("Значення")
plt.grid(True)
plt.legend()
plt.show()

print("Висновок: похідна показує швидкість зміни витрат. Після x = 2.5 витрати починають зростати.")
```
